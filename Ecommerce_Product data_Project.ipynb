```python
# step 1: Import Required libraries

import numpy as np
import pandas as pd
```

```python
# step 2: Load Dataset
data = pd.read_excel("C:/Raw data files/End to End Project/Ecommerce_Unclean_Project.xlsx")
print(data)
```

```python
# step 4: Check Missing Values

data.isnull().sum()
```

```python
# step 5: Check Duplicate Records

data.duplicated().sum()
```

```python
# Remove duplicates Records

data.drop_duplicates(inplace = True)
```

```python
# Check again duplicates records

data.duplicated().sum()
```

```python
# step 6: Replace Invalid Values
# Replace common invalid values with NaN

data.replace(["N/A", "NULL", " "], np.nan, inplace = True)
```

```python
# step 7: Remove Extra Space

data = data.apply(lambda x: x.str.strip() if x.dtype=="object" else x)
```

```python
# Sirf object (string) type ke columns ko select karke strip karein
text_cols = data.select_dtypes(include=['object']).columns
data[text_cols] = data[text_cols].apply(lambda x: x.str.strip())
```

```python
# step 8: Standardize Text Format
# Convert text into Proper Case.

data["Customer_Name"] = data["Customer_Name"].str.title()
data["City"] = data["City"].str.title()
data["State"] = data["State"].str.title()
print(data)
```

```python
# step 9: Validate Email Addresses
# Keep only valid email records.

data = data[data["Email"].str.contains("@", na = False)]
```

```python
# step 10: Convert Data Columns

data["Order_Date"] = pd.to_datetime(data["Order_Date"], errors = "coerce")
data["Delivery_Date"] = pd.to_datetime(data["Delivery_Date"], errors = "coerce")
```

```python
# step 11: Convert Numeric Columns

cols = ["Qty", "Unit_Price", "Discount"]

for c in cols:
    data[c] = pd.to_numeric(data[c], errors = "coerce")
```

```python
# step 12: Remove Invalid Quantity
# Remove records having negative or zero quantity

data = data[data["Qty"]>0]
```

```python
# step 13: Fill Missing Values

data["Discount"] = data["Discount"].fillna(0)
data["Phone"] = data["Phone"].fillna("Unknown")
data["Delivery_Date"] = data["Delivery_Date"].fillna(data["Order_Date"])
```

```python
# step 14: Feature Engineering
# Total Sales

data["Sales"] = data["Qty"] * data["Unit_Price"]
print(data)
```

```python
# Net Amount

data["Net_Amount"] = data["Sales"] - (data["Sales"] * data["Discount"]/100)
print(data)
```

```python
# Profit (Example: 20%)

data["Profit"] = data["Net_Amount"] * 0.20
print(data)
```

```python
# Order Month

data["Month"] = data["Order_Date"].dt.month_name()
print(data)
```

```python
# Order Year

data["Year"] = data["Order_Date"].dt.year
print(data)
```

```python
# Order Weekday

data["Weekday"] = data["Order_Date"].dt.day_name()
print(data)
```
